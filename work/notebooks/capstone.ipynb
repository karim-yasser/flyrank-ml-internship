## 1. Research Question

The goal of this project is to investigate whether historical content performance can help identify content items that may experience a decline in impressions.

The main research question is:

**Can machine learning use historical impressions, clicks, and average position to identify content items at risk of declining impressions in the following month?**

This project aims to support content teams by helping them prioritize items for manual review. The model is not intended to make automatic content changes or guarantee future performance.



## 2. Data

This project uses the FlyRank internship warehouse dataset, accessed through the approved Hugging Face release.

The analysis uses daily content performance data, including impressions, clicks, and average position. Historical performance is aggregated into monthly values to create features for predicting a possible decline in the following month.

The training data covers target months from April 2025 to May 2026. The final evaluation uses June 2026 data.

The training dataset contains 401,299 rows across 40 clients. The June evaluation dataset contains 92,375 rows across 40 clients.

The query-level dataset was excluded from the June prediction model because its available time window overlapped with the evaluation period.

Privacy was considered throughout the project. Client and content identifiers were not used as model features, and raw queries, private URLs, and row-level records should not be included in public outputs.



## 3. Methodology

The target variable identifies whether a content item experienced a decline in impressions. An item is labeled as declining when its impressions in the target month are less than 80% of its impressions in the previous month.

The model uses historical features:
- Previous-month impressions.
- Impressions from two months earlier.
- Previous-month clicks.
- Previous-month average position.
- Change in impressions between the two historical months.
- Log-transformed impression values.
- Clicks per impression.

A Random Forest classifier was selected for the final evaluation. It uses 100 trees, a maximum depth of 15, a minimum of 50 samples per leaf, balanced class weights, and a fixed random seed of 42.

The data was split by time rather than randomly. Model validation used March to May 2026, while June 2026 was reserved for final evaluation.

The baseline predicts that every eligible content item will experience a decline. This provides a simple comparison for checking whether the model adds useful predictive value.

Target-month impressions and identifier columns were excluded from the model features to reduce data leakage. Only historical performance information was used to construct the prediction features.



## 4. Results (vs. Baseline)

The final evaluation was performed on the June 2026 dataset, containing 92,375 content items. The observed decline rate was 68.73%.

The Random Forest model was compared with a baseline that predicts every item as declining.

| Metric | Random Forest | All-Declining Baseline |
|---|---:|---:|
| Accuracy | 56.53% | 68.73% |
| Precision | 71.34% | 68.73% |
| Recall | 61.42% | 100.00% |
| F1-score | 66.01% | 81.47% |
| ROC-AUC | 56.35% | Not applicable |

The model achieved higher precision than the baseline, but lower accuracy, recall, and F1-score. Its ROC-AUC of 56.35% indicates weak ability to distinguish between declining and non-declining items.

Therefore, the model did not outperform the baseline on accuracy or F1-score. Its scores should only be considered a possible aid for prioritizing manual reviews, not reliable predictions of future performance.



## 5. Limitations

This project has several limitations:

1. **Weak discrimination:** The model achieved a ROC-AUC of 0.5635, which indicates limited ability to distinguish between declining and non-declining content items.

2. **Baseline comparison:** The model did not outperform the all-declining baseline on accuracy or F1-score.

3. **Missed declines:** The model recall was 61.42%, meaning it did not identify every item that experienced a decline.

4. **Limited evaluation period:** The final evaluation covers June 2026 only. Results may differ in other months or datasets.

5. **Small high-risk group:** The very-high-risk group contained only 51 items, so its observed decline rate should be interpreted cautiously.

6. **No causal evidence:** The analysis does not prove that refreshing content will increase impressions.

7. **No search-engine guarantees:** The model does not predict Google's ranking algorithm or guarantee future search performance.

For these reasons, the model should be treated as an exploratory decision-support tool. Further testing on future data is needed before relying on it in a real workflow.



## 6. Ranked Recommendations

Based on the observed results, the following actions are recommended:

1. **Use the model only to prioritize manual reviews.** Content teams may use the scores to organize a review queue, but should not make automatic changes based on predictions alone.

2. **Review content context before taking action.** Editors should consider content quality, recent updates, relevance, and business priorities before deciding whether a page needs attention.

3. **Compare the model with simple alternatives.** The all-declining baseline performed better on accuracy and F1-score. Future work should test whether the model provides practical value beyond this baseline.

4. **Evaluate the model on future months.** Testing on additional time periods would help determine whether the observed performance remains consistent.

5. **Measure operational usefulness.** Track whether the prioritized queue helps reviewers find useful cases efficiently, without assuming that content updates cause impressions to recover.

These recommendations are directional and intended to support human decisions. The current results do not justify fully automated content decisions.
