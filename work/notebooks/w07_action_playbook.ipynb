## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*

### Ranked Action Queue

The action queue prioritizes content for human review using the validated signals from the previous analysis. Each recommendation includes a reason code so that a reviewer can understand why the item was prioritized.

| Priority | Reason code | Recommended action | Human interpretation |
|---|---|---|---|
| High | STALE_CONTENT | Review and consider refreshing | The content shows a staleness or refresh-related signal. |
| High | LOW_CTR | Review title and snippet | The observed CTR is lower than the selected comparison level. |
| Medium | QUICK_WIN | Prioritize for review | The content matches the baseline criteria for a potentially actionable opportunity. |
| Medium | HIGH_OPPORTUNITY | Review for optimization | The content receives a relatively high prioritization score. |

The reason codes are intended to make the recommendations easier to review and explain. They are not proof that a particular action will improve performance.

## 2. Intended use and limits

*Who uses this, for what — and where it stops being valid.*

## Intended Use

The playbook is intended to help a content team prioritize pages for human review.

It can be used to:

- identify content that matches selected review signals
- prioritize which items should be reviewed first
- provide a clear reason for each recommendation
- support consistent content-review decisions
- help allocate limited content optimization resources

The ranking is a decision-support tool. A higher-ranked item should receive attention earlier, but it does not guarantee a larger improvement in performance.

## Limitations

The results should be interpreted as observational and directional rather than causal.

The Week 5 model was trained against a rule-based target. Therefore, strong model performance can reflect how well the model reproduces the rules used to create the target rather than independent evidence of future business outcomes.

The playbook should not be treated as a guarantee that a recommended content change will increase CTR, improve ranking, or produce business value.

## 3. Human review + the no-go list

*What a person must check before acting. What should never be automated.*

## Human Review Rules

Every recommended action must be reviewed by a person before implementation.

The reviewer should check:

1. Whether the recommendation makes sense for the specific content.
2. Whether the content is actually outdated or needs refreshing.
3. Whether the recommendation matches the user's search intent.
4. Whether the page has important business, brand, or editorial constraints.
5. Whether the expected value justifies the effort required.
6. Whether the available evidence is sufficient to support the proposed action.

## No-Go List

The system should never automatically:

- publish content changes
- delete pages
- redirect pages
- change important business or factual claims
- rewrite sensitive content without review
- make final SEO decisions without human approval
- treat an observed association as proof of causation
- assume that a high score guarantees business impact
- automatically apply recommendations to every page

## 4. Monitoring / retrain triggers

*What would tell you the recommendations went stale?*

## Monitoring

The recommendations should be monitored over time to detect when the underlying patterns may have changed.

The following signals should be monitored:

- distribution of recommendation scores
- number of items receiving each reason code
- CTR distribution
- position distribution
- missing values in important features
- unusual changes in the number of recommended actions
- changes in the mix of clients or content
- changes in the distribution of important input signals

Large or unexpected changes should trigger a review of the scoring logic and data quality.

## Retraining / Revalidation Triggers

The model or scoring approach should be revalidated when:

- the target definition changes
- important features are changed or removed
- new features are introduced
- the underlying data distribution changes substantially
- recommendation patterns change unexpectedly
- performance decreases on new holdout data
- the client or content population changes significantly
- new data reveals possible leakage or validation problems

Retraining should be based on evidence from new validation data rather than performed automatically on a fixed schedule.
